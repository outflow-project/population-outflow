## 탐색적 데이터 분석

1. 데이터 분포 확인

```python
import pandas as pd
```

```python
economy = pd.read_csv('../data/processed/경제활동_전체.csv')
```

```python
economy.head()
```

```python
economy.iloc[:,2:5].corr()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#plt.pcolor(economy.iloc[:,2:5], cmap='Reds')
sns.heatmap(economy.iloc[:,2:6].corr(), annot=True)
```

* 2020년 소매와 서비스 사업체 수는 0.97 이상의 높은 선형 관계를 보인다.
* 2020년 서비스와 운송 사업체수가 0.8 이상의 높은 선형 관계를 보인다.

```python
tmp = economy.columns

tmp[tmp.str.contains('21')]
```

```python
## 나머지 연도까지 한눈에 살펴보기
#plt.figure(figsize=(30,20))

fig, ax = plt.subplots(2,2, figsize=(30,30))

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('20')]].corr(), annot=True, ax=ax[0,0])
ax[0,0].set_title('20')
ax[0,0].set_xticklabels(ax[0,0].get_xticklabels(), rotation=45)
ax[0,0].set_yticklabels(ax[0,0].get_yticklabels(), rotation=45)

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('21')]].corr(), annot=True, ax=ax[0,1])
ax[0,1].set_title('21')
ax[0,1].set_xticklabels(ax[0,1].get_xticklabels(), rotation=45)
ax[0,1].set_yticklabels(ax[0,1].get_yticklabels(), rotation=45)
   
sns.heatmap(economy.loc[:,tmp[tmp.str.contains('22')]].corr(), annot=True, ax=ax[1,0])
ax[1,0].set_title('22')
ax[1,0].set_xticklabels(ax[1,0].get_xticklabels(), rotation=45)
ax[1,0].set_yticklabels(ax[1,0].get_yticklabels(), rotation=45)

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('23')]].corr(), annot=True, ax=ax[1,1])
ax[1,1].set_title('23')
ax[1,1].set_xticklabels(ax[1,1].get_xticklabels(), rotation=45)
ax[1,1].set_yticklabels(ax[1,1].get_yticklabels(), rotation=45)


plt.subplots_adjust(hspace=0.3)

#sns.heatmap(economy.iloc[:,6:10].corr(), annot=True)
```

```python
## 나머지 연도까지 한눈에 살펴보기
#plt.figure(figsize=(30,20))

fig, ax = plt.subplots(2,2, figsize=(30,30))

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('20')]].corr(), annot=True, ax=ax[0,0])
ax[0,0].set_title('20')
ax[0,0].set_xticklabels(ax[0,0].get_xticklabels(), rotation=45)
ax[0,0].set_yticklabels(ax[0,0].get_yticklabels(), rotation=45)

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('21')]].corr(), annot=True, ax=ax[0,1])
ax[0,1].set_title('21')
ax[0,1].set_xticklabels(ax[0,1].get_xticklabels(), rotation=45)
ax[0,1].set_yticklabels(ax[0,1].get_yticklabels(), rotation=45)
   
sns.heatmap(economy.loc[:,tmp[tmp.str.contains('22')]].corr(), annot=True, ax=ax[1,0])
ax[1,0].set_title('22')
ax[1,0].set_xticklabels(ax[1,0].get_xticklabels(), rotation=45)
ax[1,0].set_yticklabels(ax[1,0].get_yticklabels(), rotation=45)

sns.heatmap(economy.loc[:,tmp[tmp.str.contains('23')]].corr(), annot=True, ax=ax[1,1])
ax[1,1].set_title('23')
ax[1,1].set_xticklabels(ax[1,1].get_xticklabels(), rotation=45)
ax[1,1].set_yticklabels(ax[1,1].get_yticklabels(), rotation=45)


plt.subplots_adjust(hspace=0.3)
```
