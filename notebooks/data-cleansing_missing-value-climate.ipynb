```python
## 인접 시군의 평균으로 결측값이 있다면 채웁니다.
# def fill_with_neighbors_mean(row, target_col, neigh_dict, df):
#     if pd.isnull(row[target_col]):
#         neighbors = neigh_dict.get(row['SGG_NAME'], [])
#         neigh_vals = df[df['SGG_NAME'].isin(neighbors)][target_col]
#         mean_val = neigh_vals.mean()
#         return round(mean_val,1)
#     else:
#         return row[target_col]

def fill_with_neighbors_mean(row, target_col, neigh_dict, df):
    if pd.isnull(row[target_col]):
        neighbors = neigh_dict.get(row['SGG_NAME'], [])
        neigh_vals = df[df['SGG_NAME'].isin(neighbors)][target_col]
        # 결측이 아닌 값만 필터링
        valid_values = neigh_vals.dropna()
        if len(valid_values) > 0:
            return round(valid_values.mean(),1)
        else:
            # 모든 이웃값이 결측이면 원래 값 유지(NaN)
            return row[target_col]
    else:
        return row[target_col]
```

```python
for target_col in climate_edit.columns:
    climate_edit[target_col] = climate_edit.apply(
        lambda row: fill_with_neighbors_mean(row, target_col, adjacent_dict, climate_edit), axis=1
    )
```
