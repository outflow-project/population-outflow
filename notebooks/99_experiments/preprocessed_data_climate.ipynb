## 기후데이터 기초전처리
* 같은 시군구 내 관측소 값을 각 해마다 평균내어 하나로 처리

```python
import pandas as pd
```

```python
climate_sudo = pd.read_csv('../data/raw/기후데이터_수도권0811.csv', encoding='cp949')
```

```python
climate_sudo.head(10)
```

```python
climate_sudo_mean = climate_sudo.groupby(['시군구별(1)','시군구별(2)']).mean().round(1)
```

```python
print(climate_sudo_mean.shape)
```

```python
climate_sudo_mean.tail()
```

```python
climate_sudo_mean.to_csv('../data/climate(기후데이터)/기후데이터_컬럼명변경_수도권.csv', encoding='utf-8')
```
