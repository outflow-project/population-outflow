## 보건 데이터 확인

```python
import glob
import os

health_dir = '../data/보건/'

for hos in glob.glob('../data/보건/*'):
    tmp = pd.read_csv(hos)
    print(os.path.basename(hos), ':')
    
    print('- 의 개수:')

    for col in tmp.columns:
        print(f"{col}: {len(tmp.loc[tmp[col] == '-', :])}")
    
    print('결측치 개수: ')
    print(tmp.isnull().sum())
    print(list(tmp.loc[tmp.isnull().any(axis=1),'SGG_NAME']))
    print('='*50)
```

```python
welf = pd.read_csv(health_dir + '보건복지시설수_전국.csv')
```

```python
welf.loc[welf.isnull().any(axis=1),:]
```

```python
welf2 = welf.copy()
welf2 = welf2.fillna(0)
```

```python
welf2.loc[welf.isnull().any(axis=1),:]
```

```python
print(welf2.isnull().sum())
```

```python
for col in welf2.columns:
    print(col, len(welf2.loc[welf2[col] == '-', :]))
```

```python
per1000 = pd.read_csv(health_dir + '천명당의료기관병상수_전국.csv')
per1000.head()
```

```python
per1000.isnull().sum()
```

```python
per1000.loc[per1000.isnull().any(axis=1),:]
per1000_zero = per1000.fillna(0)
```

```python
per1000_zero.isnull().sum()
```

```python
for col in per1000_zero.columns:
    print(col, len(per1000_zero.loc[per1000_zero[col] == '-',:]))
```

```python
totbed = pd.read_csv(health_dir + '총병상수_전국.csv')
totbed.isnull().sum()
```

```python
totbed.loc[totbed.isnull().any(axis=1),:]
## 0으로 채우기
totbed_zero = totbed.fillna(0)
```

```python
totbed_zero.loc[totbed.isnull().any(axis=1),:]
```

```python
totbed_zero.isnull().sum()
```

```python
for col in totbed_zero.columns:
    print(col, len(totbed_zero.loc[totbed_zero[col] == '-',:]))
```

```python
## to_csv
welf2.to_csv('보건복지시설수.csv', index=False)
per1000_zero.to_csv('천명당의료기관병상수.csv', index=False)
totbed_zero.to_csv('총병상수.csv', index=False)
```
